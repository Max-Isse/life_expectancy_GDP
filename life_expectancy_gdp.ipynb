# **Imports**


```python
#%pip install matplotlib
#%pip install seaborn
#%pip install scipy
#%pip install statsmodels

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr, spearmanr
import statsmodels.api as sm
```

# **Visualization settings**

```python
sns.set_context("talk")
sns.set_style("whitegrid")
```


# **Data Loading**


```python
who_life = pd.read_csv("../data/who_life_expectancy.csv")
gdp = pd.read_csv("../data/gdp_per_capita.csv")
```


# **Initial Inspection**


```python
who_life.head(), who_life.info()
gdp.head(), gdp.info()
```


# **Data Cleaning**

## ***Standardize column names***

```python
who_life.columns = who_life.columns.str.lower()
gdp.columns = gdp.columns.str.lower()
```

## ***Rename key variables if needed***

```python
who_life.rename(columns={"life expectancy (years)": "life_expectancy"}, inplace=True)
```

## ***Merge datasets***

df = pd.merge(
    who_life,
    gdp,
    on=["country", "year"],
    how="inner"
)


## ***Remove missing or invalid values***

```python
df = df.dropna(subset=["life_expectancy", "gdp_per_capita"])
df = df[df["gdp_per_capita"] > 0]
```


# **Exploratory Data Analysis**


```python
df.describe()
```

## ***Correlations***

```python
pearson_corr, _ = pearsonr(df["gdp_per_capita"], df["life_expectancy"])
spearman_corr, _ = spearmanr(df["gdp_per_capita"], df["life_expectancy"])

print(f"Pearson correlation: {pearson_corr:.3f}")
print(f"Spearman correlation: {spearman_corr:.3f}")
```


## ***Visualization***


```python
plt.figure()
sns.scatterplot(
    data=df,
    x="gdp_per_capita",
    y="life_expectancy",
    alpha=0.6
)
plt.xscale("log")
plt.title("Life Expectancy vs GDP per Capita (Log Scale)")
plt.xlabel("GDP per Capita (log scale)")
plt.ylabel("Life Expectancy (years)")
plt.show()
```


## ***Regression Analysis***


```python
df["log_gdp"] = np.log(df["gdp_per_capita"])

X = sm.add_constant(df["log_gdp"])
y = df["life_expectancy"]

model = sm.OLS(y, X).fit()
model.summary()

```


# **Key Insights**


```python
"""
- Relationship is positive but nonlinear
- Strong gains at low income levels
- Diminishing returns at higher GDP
- Significant cross-country heterogeneity
"""
```


# **Conclusion**


```python
"""
This analysis confirms that while economic growth is associated
with longer life expectancy, income alone does not fully explain
health outcomes. Institutional quality, healthcare access, and
inequality play critical roles.
"""
```
